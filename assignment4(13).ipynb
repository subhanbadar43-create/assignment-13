```python
def calculatemobilebill(package, extragb):
    prices = {
        "Basic": 1000,
        "Standard": 1500,
        "Premium": 2500
    }
    freegbmap = {
        "Basic": 0,
        "Standard": 2,
        "Premium": 5
    }
    packageprice = prices[package]
    freegb = freegbmap[package]
    chargedgb = max(0, extragb - freegb)
    extracharges = chargedgb * 100
    finalbill = packageprice + extracharges
    print(f"Package: {package}")
    print(f"Extra Data: {extragb} GB")
    print()
    print(f"Package Price: {packageprice}")
    print(f"Free Data: {freegb} GB")
    print(f"Charged Data: {chargedgb} GB")
    print(f"Extra Charges: {extracharges}")
    print()
    print(f"Final Bill: {finalbill}")
    return finalbill
calculatemobilebill("Standard", 5)
```
